# Example: Querying Databricks from VS Code

This notebook shows how to use **SQL magic commands** to query Databricks tables directly from VS Code.

### Prerequisites
1. The **Databricks VS Code extension** must be connected (check the sidebar)
2. A compute resource must be attached (serverless or cluster)

### How it works
- Use `%sql` at the top of a cell to run a single SQL statement
- Results display inline as a table below the cell
- This is great for exploring data, validating transformations, and ad-hoc analysis

```python
%sql
-- Example: Query the unified AR invoice table
-- Change the catalog/schema to match your domain
SELECT *
FROM silver_astra.accounts_receivable.astra_ar_invoice
LIMIT 100
```

```python
%sql
-- Example: Quick record count by company
SELECT
    `Company Name`,
    `System`,
    COUNT(*) as record_count,
    SUM(`Invoice Amount`) as total_invoiced
FROM silver_astra.accounts_receivable.astra_ar_invoice
GROUP BY `Company Name`, `System`
ORDER BY total_invoiced DESC
```

### Tips
- Use `silver_astra_dev` instead of `silver_astra` when working in the dev environment
- Column names in silver_astra tables use Title Case with backticks: `` `Company Name` ``
- Column names in silver_opco tables use snake_case: `company_name`
- You can also use `%python` cells for more complex analysis with pandas/spark
